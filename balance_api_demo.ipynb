# Balance API Demonstration

This notebook demonstrates how to interact with the deployed Balance API on Hugging Face Spaces.

**API Endpoint:** https://nsomabalint-bank-api.hf.space/api/balance

## Test Account 1: John Doe (PIN: 1234)

Expected: 15,420.50 USD

```python
%%bash
curl -X POST https://nsomabalint-bank-api.hf.space/api/balance \
  -H "Content-Type: application/json" \
  -d '{"pin": "1234"}' | jq
```
